```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'text' : ['people watch campusx', 'campusx watch campusx', 'people write comment', 'campusx write comment'], 'output' : [1,1,0,0]})
```

```python
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf.fit_transform(df['text']).toarray()   # since it returns sparse matrix, so for converting it into 'numpy_array' we use "toarray()"
```

```python
# for printing the idf value of the constant-

print(tfidf.idf_)
print(tfidf.get_feature_names_out())
```
